```python
# Program 1: Python program to create SHA-256 hash of a given string

import hashlib

def create_hash(string):
    # Create a hash object using SHA-256 algorithm
    hash_object = hashlib.sha256()

    # Convert the string to bytes and update the hash object
    hash_object.update(string.encode('utf-8'))

    # Get the hexadecimal representation of the hash
    hash_string = hash_object.hexdigest()

    # Return the hash string
    return hash_string
```

```python
# Program 3: Solving Cryptographic Puzzle using Nonce

import hashlib

def find_nonce(input_string, num_zeros):
    nonce = 0
    hash_prefix = '0' * num_zeros

    while True:
        # Concatenate input string and nonce
        hash_string = input_string + str(nonce)

        # Calculate SHA-256 hash
        hash_object = hashlib.sha256(hash_string.encode('utf-8'))
        hash_code = hash_object.hexdigest()

        # Check whether hash starts with required zeros
        if hash_code.startswith(hash_prefix):
            print("\nHash Found:", hash_code)
            return nonce

        nonce += 1


# Get user input
input_string = input("Enter a string: ")
num_zeros = int(input("Enter the number of leading zeros required: "))

# Find the nonce
expected_nonce = find_nonce(input_string, num_zeros)

# Display result
print("\nName: Jai Udasi")
print("Input String:", input_string)
print("Leading Zeros Required:", num_zeros)
print("Expected Nonce:", expected_nonce)
```

**Program 4: Generating Merkle Tree for given set of Transactions**

```python
# Program 4: Constructing a Merkle Tree Root Hash
# Name: Jai Udasi

import hashlib

# ---------------------------------------------------------
# FUNCTION 1: Generate SHA-256 Hash
# ---------------------------------------------------------
def hash_data(data):
    return hashlib.sha256(data.encode("utf-8")).hexdigest()


# ---------------------------------------------------------
# FUNCTION 2: Shorten Hash for Tree Display
# ---------------------------------------------------------
def short_hash(hash_value):
    return hash_value[:10] + "..."


# ---------------------------------------------------------
# FUNCTION 3: Print Intermediate Hashes
# ---------------------------------------------------------
def print_hash_levels(levels):

    print("\n" + "=" * 90)
    print("              INTERMEDIATE MERKLE TREE HASHES")
    print("=" * 90)

    for level_no, level in enumerate(levels):

        if level_no == 0:
            print("\nLEVEL 0 - TRANSACTION HASHES")
        else:
            print(f"\nLEVEL {level_no}")

        for i, node in enumerate(level):

            print(f"  H{i + 1}: {node}")


# ---------------------------------------------------------
# FUNCTION 4: Build Merkle Tree
# ---------------------------------------------------------
def build_merkle_tree(transactions):

    # Hash every transaction first
    current_level = []

    for transaction in transactions:
        current_level.append(hash_data(transaction))

    # Store all levels of the tree
    levels = [current_level.copy()]

    while len(current_level) > 1:

        # If number of hashes is odd,
        # duplicate the last hash
        if len(current_level) % 2 != 0:
            current_level.append(current_level[-1])

        next_level = []

        # Combine hashes in pairs
        for i in range(0, len(current_level), 2):

            combined = current_level[i] + current_level[i + 1]

            parent_hash = hash_data(combined)

            next_level.append(parent_hash)

        current_level = next_level

        levels.append(current_level.copy())

    return current_level[0], levels


# ---------------------------------------------------------
# FUNCTION 5: Print ASCII MERKLE TREE
# ---------------------------------------------------------
def print_merkle_tree(transactions, levels):

    print("\n\n" + "=" * 90)
    print("                    MERKLE TREE")
    print("=" * 90)

    # -----------------------------------------------------
    # Print levels from ROOT to TRANSACTIONS
    # -----------------------------------------------------

    root_level = levels[-1]

    print("\n                         ┌──────────────────────┐")
    print("                         │     MERKLE ROOT      │")
    print(f"                         │ {short_hash(root_level[0]):^20} │")
    print("                         └──────────┬───────────┘")

    # Work backwards from second-last level
    for level_index in range(len(levels) - 2, -1, -1):

        current_level = levels[level_index]

        print("                                    │")

        # Print connecting line
        print("                     ┌──────────────┴──────────────┐")

        # Display nodes
        if len(current_level) == 1:

            print(
                f"                     │ "
                f"[{short_hash(current_level[0])}]"
                f" │"
            )

        else:

            # Create pairs for display
            for i in range(0, len(current_level), 2):

                left = short_hash(current_level[i])

                if i + 1 < len(current_level):
                    right = short_hash(current_level[i + 1])
                else:
                    right = left

                print(
                    f"                  [{left}]       [{right}]"
                )

        # Separator between levels
        if level_index > 0:
            print("                              │")

    # -----------------------------------------------------
    # Print Transactions at Bottom
    # -----------------------------------------------------

    print("\n")
    print("-" * 90)
    print("                    TRANSACTIONS")
    print("-" * 90)

    for i, transaction in enumerate(transactions):

        print(
            f"T{i + 1} -> {transaction}"
        )

    print("-" * 90)


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

print("\n" + "=" * 90)
print("                  MERKLE TREE GENERATOR")
print("                         Jai Udasi")
print("=" * 90)


# ---------------------------------------------------------
# USER INPUT
# ---------------------------------------------------------

while True:

    try:
        number = int(
            input("\nEnter number of transactions: ")
        )

        if number > 0:
            break

        print("Please enter a number greater than 0.")

    except ValueError:
        print("Please enter a valid integer.")


transactions = []

print("\nEnter your transactions:")

for i in range(number):

    transaction = input(f"T{i + 1}: ")

    transactions.append(transaction)


# ---------------------------------------------------------
# DISPLAY TRANSACTION INFORMATION
# ---------------------------------------------------------

print("\n" + "=" * 90)
print("                    INPUT TRANSACTIONS")
print("=" * 90)

for i, transaction in enumerate(transactions):

    print(f"T{i + 1}: {transaction}")


# ---------------------------------------------------------
# HASH EACH TRANSACTION
# ---------------------------------------------------------

print("\n" + "=" * 90)
print("                  TRANSACTION HASHES")
print("=" * 90)

for i, transaction in enumerate(transactions):

    transaction_hash = hash_data(transaction)

    print(f"\nT{i + 1}: {transaction}")
    print(f"Hash: {transaction_hash}")


# ---------------------------------------------------------
# BUILD MERKLE TREE
# ---------------------------------------------------------

merkle_root, levels = build_merkle_tree(transactions)


# ---------------------------------------------------------
# PRINT INTERMEDIATE HASHES
# ---------------------------------------------------------

print_hash_levels(levels)


# ---------------------------------------------------------
# PRINT ASCII MERKLE TREE
# ---------------------------------------------------------

print_merkle_tree(transactions, levels)


# ---------------------------------------------------------
# FINAL RESULT
# ---------------------------------------------------------

print("\n" + "=" * 90)
print("                       FINAL RESULT")
print("=" * 90)

print("Name              :", "Jai Udasi")
print("Number of Transactions:", number)
print("Merkle Root       :", merkle_root)

print("=" * 90)
```
